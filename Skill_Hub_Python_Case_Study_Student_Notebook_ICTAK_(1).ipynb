**Words you will meet**

| Word | Meaning |
|---|---|
| Category | The kind of learner: `student`, `govt` (government employee) or `corporate` |
| Base fee | The full price of a course before any discount |
| Discount | An amount taken off the price. Here it is a percentage of the base fee |
| Coupon | A code that gives an extra reduction, e.g. `EARLY500` |
| GST | A tax of 18% added to the final fee |
| Instalment | One of several smaller payments that together make the full amount |
| Seat | One place in a course. Every course has a limited number of seats |
| Enrol / reject | To **enrol** is to accept a learner. To **reject** is to refuse, with a reason |

## 4 · The rules of Skill Hub 📋

| Question | Rule |
|---|---|
| Who can join? | Learners who are **18 or older** and whose category is `student`, `govt` or `corporate` |
| Discount | **student 20%** · **govt 15%** · **corporate 0%**, taken off the base fee |
| Coupon `EARLY500` | Takes **Rs 500** off, but **only if** the fee after the category discount is **Rs 6000 or more** |
| Coupon `FRIEND5` | Takes **5%** off the fee after the category discount |
| Any other coupon | Not valid, so ignore it. Coupons can be typed in any capital or small letters |
| GST | Add **18%** to the final fee |
| Seats | A course with no seats left cannot take another learner |

**Worked example.** Anil is a student booking Python Basics (base fee Rs 8,000):

- Student discount: 20% of 8,000 = 1,600, so the fee after discount = **6,400**
- GST: 18% of 6,400 = 1,152, so the total to pay = **7,552**
